```python
import os
import time
import cv2
import torch
import requests
import json

from nanodet.data.batch_process import stack_batch_img
from nanodet.data.collate import naive_collate
from nanodet.data.transform import Pipeline
from nanodet.model.arch import build_model
from nanodet.util import Logger, cfg, load_config, load_model_weight
from nanodet.util.path import mkdir

# Azure OCR Configuration
AZURE_OCR_ENDPOINT = "https://vinit2.cognitiveservices.azure.com"
AZURE_SUBSCRIPTION_KEY = "2hdjCP2joPTpL8AjsME8gWP8P8I7LA4rk6juwzk1BkmhgU4tohLlJQQJ99BCACGhslBXJ3w3AAAFACOGs68y"

# Allowed Image Extensions
image_ext = [".jpg", ".jpeg", ".webp", ".bmp", ".png"]

class Predictor:
    def __init__(self, cfg, model_path, logger, device="cuda:0", score_thres=0.5):
        self.cfg = cfg
        self.device = device
        self.score_thres = score_thres
        model = build_model(cfg.model)
        ckpt = torch.load(model_path, map_location=lambda storage, loc: storage)
        load_model_weight(model, ckpt, logger)
        self.model = model.to(device).eval()
        self.pipeline = Pipeline(cfg.data.val.pipeline, cfg.data.val.keep_ratio)

    def inference(self, img):
        img_info = {"id": 0, "file_name": os.path.basename(img) if isinstance(img, str) else None}
        img = cv2.imread(img) if isinstance(img, str) else img
        img_info.update({"height": img.shape[0], "width": img.shape[1]})
        meta = dict(img_info=img_info, raw_img=img, img=img)
        meta = self.pipeline(None, meta, self.cfg.data.val.input_size)
        meta["img"] = torch.from_numpy(meta["img"].transpose(2, 0, 1)).to(self.device)
        meta = naive_collate([meta])
        meta["img"] = stack_batch_img(meta["img"], divisible=32)

        with torch.no_grad():
            results = self.model.inference(meta)
        
        return meta, results

    def visualize(self, dets, meta, class_names):
        return self.model.head.show_result(meta["raw_img"][0], dets, class_names, score_thres=self.score_thres, show=False)

    def extract_text_with_azure_ocr(self, image_path):
        """
        Send image to Azure OCR and return detected text.
        """
        headers = {
            "Ocp-Apim-Subscription-Key": AZURE_SUBSCRIPTION_KEY,
            "Content-Type": "application/octet-stream"
        }
        
        with open(image_path, "rb") as image_file:
            response = requests.post(AZURE_OCR_ENDPOINT, headers=headers, data=image_file)
        
        if response.status_code != 202:
            print(f"Error: {response.status_code}, {response.text}")
            return None

        # Get the operation location from the response
        operation_url = response.headers["Operation-Location"]

        # Poll for results
        time.sleep(2)  # Give Azure OCR some time to process
        for _ in range(10):  # Try multiple times in case processing takes time
            result_response = requests.get(operation_url, headers={"Ocp-Apim-Subscription-Key": AZURE_SUBSCRIPTION_KEY})
            result_json = result_response.json()

            if result_json.get("status") == "succeeded":
                text_lines = []
                for read_result in result_json["analyzeResult"]["readResults"]:
                    for line in read_result["lines"]:
                        text_lines.append(line["text"])
                return " ".join(text_lines)
            time.sleep(1)

        print("Azure OCR processing timeout.")
        return None


def get_image_list(path):
    return [os.path.join(maindir, filename) for maindir, _, filelist in os.walk(path) for filename in filelist if os.path.splitext(filename)[1] in image_ext] if os.path.isdir(path) else [path]


def run_inference_for_image(config_path, model_path, image_path, save_result=False, save_dir='./demo_results', score_thres=0.5):
    load_config(cfg, config_path)
    logger = Logger(local_rank=0, use_tensorboard=False)
    predictor = Predictor(cfg, model_path, logger, device="cuda:0", score_thres=score_thres)

    image_names = get_image_list(image_path)
    image_names.sort()

    if save_result:
        save_folder = os.path.join(save_dir, time.strftime("%Y_%m_%d_%H_%M_%S", time.localtime()))
        mkdir(local_rank=0, path=save_folder)

    result_images = []
    for image_name in image_names:
        meta, res = predictor.inference(image_name)
        result_image = predictor.visualize(res[0], meta, cfg.class_names)

        # Run Azure OCR on the detected image
        detected_text = predictor.extract_text_with_azure_ocr(image_name)
        print(f"Detected text in {image_name}: {detected_text}")

        if save_result:
            save_file_name = os.path.join(save_folder, os.path.basename(image_name))
            cv2.imwrite(save_file_name, result_image)

        result_images.append(result_image)
    return result_images


# Example usage
config_path = 'config/nanodet-plus-m_416-yolo.yml'
model_path = 'workspace/nanodet-plus-m_416/model_best/nanodet_model_best.pth'
image_path = 'License-Plate.v7-2.coco-mmdetection/test/Fiat-Grande-Punto-525629c_jpg_0000_0293_0434_0216_0057_png.rf.232c0089c3bc1e0b44b1a10e7b3c5678.jpg'
result_images = run_inference_for_image(config_path, model_path, image_path, save_result=True, save_dir='Results', score_thres=0.5)

# Display results
import matplotlib.pyplot as plt
if result_images:
    plt.imshow(cv2.cvtColor(result_images[0], cv2.COLOR_BGR2RGB))
    plt.axis('off')
    plt.show()
else:
    print("No result images available.")
```

```python
import os
import time
from azure.cognitiveservices.vision.computervision import ComputerVisionClient
from azure.cognitiveservices.vision.computervision.models import OperationStatusCodes
from msrest.authentication import CognitiveServicesCredentials

# ✅ Load API Key & Endpoint securely from environment variables
api_key = os.getenv("AZURE_API_KEY")
endpoint = os.getenv("AZURE_ENDPOINT")

if not api_key or not endpoint:
    raise ValueError("Azure API Key or Endpoint is missing! Set environment variables properly.")

# ✅ Initialize the Computer Vision Client
client = ComputerVisionClient(endpoint, CognitiveServicesCredentials(api_key))

def extract_text_from_image(image_path):
    """Extracts text from an image using Azure OCR."""
    try:
        # ✅ Open image as binary stream
        with open(image_path, "rb") as image_file:
            response = client.read_in_stream(image_file, raw=True)  # Add raw=True

        # ✅ Check if response is None
        if response is None:
            return "Error: No response received from Azure OCR API."

        # ✅ Extract operation ID correctly
        operation_location = response.headers.get("Operation-Location")
        if not operation_location:
            return "Error: Operation-Location header missing in response."

        operation_id = operation_location.split("/")[-1]

        # ✅ Wait for OCR to complete (Max 30s timeout)
        print("Processing OCR... Please wait.")
        max_wait_time = 30  # Max wait time in seconds
        start_time = time.time()

        while True:
            result = client.get_read_result(operation_id)
            if result.status in [OperationStatusCodes.succeeded, OperationStatusCodes.failed]:
                break
            if time.time() - start_time > max_wait_time:
                return "Error: OCR processing timed out."
            time.sleep(2)  # Delay to avoid excessive API calls

        # ✅ Extract and return text
        extracted_text = []
        if result.status == OperationStatusCodes.succeeded:
            for page in result.analyze_result.read_results:
                for line in page.lines:
                    extracted_text.append(line.text)

        return "\n".join(extracted_text) if extracted_text else "No text found in the image."

    except Exception as e:
        return f"Error: {str(e)}"


# Example usage
image_path = "Test/PXL_20250310_062331085.jpg"  # Replace with actual image path
text_output = extract_text_from_image(image_path)
print("Extracted Text:\n", text_output)
```

```python
import os
import time
import cv2
import torch
import json
import requests

from nanodet.data.batch_process import stack_batch_img
from nanodet.data.collate import naive_collate
from nanodet.data.transform import Pipeline
from nanodet.model.arch import build_model
from nanodet.util import Logger, cfg, load_config, load_model_weight
from nanodet.util.path import mkdir

image_ext = [".jpg", ".jpeg", ".webp", ".bmp", ".png"]

# Azure OCR Configuration from Environment Variables
AZURE_OCR_ENDPOINT = os.getenv("AZURE_OCR_ENDPOINT")
AZURE_OCR_KEY = os.getenv("AZURE_OCR_KEY")

CONFIDENCE_THRESHOLD = 0.50

def azure_ocr(image_path):
    with open(image_path, "rb") as image_file:
        image_data = image_file.read()
    
    headers = {
        "Ocp-Apim-Subscription-Key": AZURE_OCR_KEY,
        "Content-Type": "application/octet-stream"
    }
    params = {
        "language": "en",
        "detectOrientation": "true"
    }
    response = requests.post(f"{AZURE_OCR_ENDPOINT}/vision/v3/ocr", headers=headers, params=params, data=image_data)
    
    if response.status_code == 200:
        ocr_result = response.json()
        extracted_text = " ".join(
            word["text"] for region in ocr_result.get("regions", []) for line in region.get("lines", []) for word in line.get("words", [])
        )
        return extracted_text
    else:
        print("Azure OCR error:", response.text)
        return None

def run_inference_for_image(config_path, model_path, image_path, save_result=False, save_dir='./demo_results', score_thres=0.5):
    load_config(cfg, config_path)
    logger = Logger(local_rank=0, use_tensorboard=False)
    predictor = Predictor(cfg, model_path, logger, device="cuda:0", score_thres=score_thres)
    
    image_names = get_image_list(image_path)
    image_names.sort()

    if save_result:
        save_folder = os.path.join(save_dir, time.strftime("%Y_%m_%d_%H_%M_%S", time.localtime()))
        mkdir(local_rank=0, path=save_folder)
    
    result_images = []
    for image_name in image_names:
        meta, res = predictor.inference(image_name)
        result_image = predictor.visualize(res[0], meta, cfg.class_names)
        
        if save_result:
            save_file_name = os.path.join(save_folder, os.path.basename(image_name))
            cv2.imwrite(save_file_name, result_image)
        
        result_images.append(result_image)
        
        # Run Azure OCR on detected region (assuming license plate detection)
        license_plate_text = azure_ocr(image_name)
        if license_plate_text:
            print(f"Detected License Plate Text: {license_plate_text}")
        else:
            print("No license plate text detected.")
    
    return result_images

class Predictor:
    def __init__(self, cfg, model_path, logger, device="cuda:0", score_thres=0.5):
        self.cfg = cfg
        self.device = device
        self.score_thres = score_thres
        model = build_model(cfg.model)
        ckpt = torch.load(model_path, map_location=lambda storage, loc: storage)
        load_model_weight(model, ckpt, logger)
        self.model = model.to(device).eval()
        self.pipeline = Pipeline(cfg.data.val.pipeline, cfg.data.val.keep_ratio)
    
    def inference(self, img):
        img_info = {"id": 0, "file_name": os.path.basename(img) if isinstance(img, str) else None}
        img = cv2.imread(img) if isinstance(img, str) else img
        img_info.update({"height": img.shape[0], "width": img.shape[1]})
        meta = dict(img_info=img_info, raw_img=img, img=img)
        meta = self.pipeline(None, meta, self.cfg.data.val.input_size)
        meta["img"] = torch.from_numpy(meta["img"].transpose(2, 0, 1)).to(self.device)
        meta = naive_collate([meta])
        meta["img"] = stack_batch_img(meta["img"], divisible=32)
        with torch.no_grad():
            results = self.model.inference(meta)
        return meta, results
    
    def visualize(self, dets, meta, class_names):
        return self.model.head.show_result(meta["raw_img"][0], dets, class_names, score_thres=self.score_thres, show=False)

def get_image_list(path):
    return [os.path.join(maindir, filename) for maindir, _, filelist in os.walk(path) for filename in filelist if os.path.splitext(filename)[1] in image_ext] if os.path.isdir(path) else [path]

# Example usage
config_path = 'config/nanodet-plus-m_416-yolo.yml'
model_path = 'workspace/nanodet-plus-m_416/model_best/nanodet_model_best.pth'
image_path = 'Test/PXL_20250310_062246240.jpg'
result_images = run_inference_for_image(config_path, model_path, image_path, save_result=True, save_dir='Results', score_thres=0.5)

# Display results
import matplotlib.pyplot as plt
if result_images:
    plt.imshow(cv2.cvtColor(result_images[0], cv2.COLOR_BGR2RGB))
    plt.axis('off')
    plt.show()
else:
    print("No result images available.")
```
