```python
import os
import time
import cv2
import torch
import numpy as np
import matplotlib.pyplot as plt
from nanodet.data.batch_process import stack_batch_img
from nanodet.data.collate import naive_collate
from nanodet.data.transform import Pipeline
from nanodet.model.arch import build_model
from nanodet.util import Logger, cfg, load_config, load_model_weight
from nanodet.util.path import mkdir
from azure.cognitiveservices.vision.computervision import ComputerVisionClient
from azure.cognitiveservices.vision.computervision.models import OperationStatusCodes
from msrest.authentication import CognitiveServicesCredentials
from io import BytesIO

image_ext = [".jpg", ".jpeg", ".webp", ".bmp", ".png"]
# Confidence Threshold for Detection
CONFIDENCE_THRESHOLD = 0.50

# Initialize Azure OCR Client
def initialize_azure_client():
    """Initialize and return Azure Computer Vision client"""
    api_key = os.getenv("AZURE_API_KEY")
    endpoint = os.getenv("AZURE_ENDPOINT")
    if not api_key or not endpoint:
        raise ValueError("Azure API Key or Endpoint is missing! Set environment variables properly.")
    return ComputerVisionClient(endpoint, CognitiveServicesCredentials(api_key))

def extract_text_from_image(client, image):
    """
    Extracts text from an image using Azure OCR.
    
    Args:
        client: Azure Computer Vision client
        image: Image as numpy array (BGR format)
    
    Returns:
        Extracted text as string
    """
    try:
        # Convert numpy array to bytes
        is_success, buffer = cv2.imencode(".jpg", image)
        if not is_success:
            return "Error: Failed to encode image."
        
        image_bytes = buffer.tobytes()
        
        # Convert bytes to BytesIO object (file-like object)
        from io import BytesIO
        image_stream = BytesIO(image_bytes)
        
        # Send image to Azure OCR
        response = client.read_in_stream(image_stream, raw=True)
        
        # Rest of the function remains the same
        # ...
        
        # Check if response is None
        if response is None:
            return "Error: No response received from Azure OCR API."
        
        # Extract operation ID correctly
        operation_location = response.headers.get("Operation-Location")
        if not operation_location:
            return "Error: Operation-Location header missing in response."
        
        operation_id = operation_location.split("/")[-1]
        
        # Wait for OCR to complete (Max 30s timeout)
        max_wait_time = 30  # Max wait time in seconds
        start_time = time.time()
        while True:
            result = client.get_read_result(operation_id)
            if result.status in [OperationStatusCodes.succeeded, OperationStatusCodes.failed]:
                break
            if time.time() - start_time > max_wait_time:
                return "Error: OCR processing timed out."
            time.sleep(2)  # Delay to avoid excessive API calls
        
        # Extract and return text
        extracted_text = []
        if result.status == OperationStatusCodes.succeeded:
            for page in result.analyze_result.read_results:
                for line in page.lines:
                    extracted_text.append(line.text)
        
        return "\n".join(extracted_text) if extracted_text else "No text found in the image."
    
    except Exception as e:
        return f"Error: {str(e)}"

def crop_detected_objects(image, dets, class_names, score_thres=0.5):
    """
    Crop and return detected objects from the image
    Args:
        image: original image
        dets: detection results
        class_names: list of class names
        score_thres: score threshold for filtering detections
    Returns:
        List of tuples (cropped_image, class_name, score, bbox)
    """
    cropped_objects = []
    
    for label in range(len(class_names)):
        for bbox in dets[label]:
            score = bbox[-1]
            if score > score_thres:
                x0, y0, x1, y1 = [int(i) for i in bbox[:4]]
                # Ensure box is within image boundaries
                x0 = max(0, x0)
                y0 = max(0, y0)
                x1 = min(image.shape[1], x1)
                y1 = min(image.shape[0], y1)
                
                # Crop the object
                cropped = image[y0:y1, x0:x1]
                
                if cropped.size > 0:  # Check if crop is valid
                    cropped_objects.append((
                        cropped, 
                        class_names[label], 
                        score, 
                        (x0, y0, x1, y1)
                    ))
    
    return cropped_objects

def run_inference_for_image(config_path, model_path, image_path, azure_client=None, save_result=False, save_dir='./demo_results', score_thres=0.5):
    load_config(cfg, config_path)
    logger = Logger(local_rank=0, use_tensorboard=False)
    predictor = Predictor(cfg, model_path, logger, device="cuda:0", score_thres=score_thres)
    
    image_names = get_image_list(image_path)
    image_names.sort()
    if save_result:
        save_folder = os.path.join(save_dir, time.strftime("%Y_%m_%d_%H_%M_%S", time.localtime()))
        mkdir(local_rank=0, path=save_folder)
        
        # Create subfolder for crops
        crop_folder = os.path.join(save_folder, "crops")
        mkdir(local_rank=0, path=crop_folder)
    
    result_images = []
    all_cropped_objects = []
    
    for image_name in image_names:
        meta, res = predictor.inference(image_name)
        raw_img = meta["raw_img"][0]
        result_image = predictor.visualize(res[0], meta, cfg.class_names)
        
        # Crop detected objects
        cropped_objects = crop_detected_objects(raw_img, res[0], cfg.class_names, score_thres)
        
        # Process OCR if Azure client is provided
        if azure_client:
            cropped_objects_with_ocr = []
            for crop, class_name, score, bbox in cropped_objects:
                ocr_text = extract_text_from_image(azure_client, crop)
                cropped_objects_with_ocr.append((
                    crop, 
                    class_name, 
                    score, 
                    bbox, 
                    ocr_text
                ))
            all_cropped_objects.append((os.path.basename(image_name), cropped_objects_with_ocr))
        else:
            all_cropped_objects.append((os.path.basename(image_name), cropped_objects))
        
        if save_result:
            # Save the image with bounding boxes
            save_file_name = os.path.join(save_folder, os.path.basename(image_name))
            cv2.imwrite(save_file_name, result_image)
            
            # Save cropped objects
            for i, item in enumerate(all_cropped_objects[-1][1]):
                if azure_client:
                    crop, class_name, score, _, ocr_text = item
                    # Save OCR text to file
                    text_filename = f"{os.path.splitext(os.path.basename(image_name))[0]}_{class_name}_{i}_{score:.2f}.txt"
                    text_path = os.path.join(crop_folder, text_filename)
                    with open(text_path, 'w') as f:
                        f.write(ocr_text)
                else:
                    crop, class_name, score, _ = item
                
                crop_filename = f"{os.path.splitext(os.path.basename(image_name))[0]}_{class_name}_{i}_{score:.2f}.jpg"
                crop_path = os.path.join(crop_folder, crop_filename)
                cv2.imwrite(crop_path, crop)
        
        result_images.append(result_image)
    
    return result_images, all_cropped_objects

class Predictor:
    def __init__(self, cfg, model_path, logger, device="cuda:0", score_thres=0.5):
        self.cfg = cfg
        self.device = device
        self.score_thres = score_thres
        model = build_model(cfg.model)
        ckpt = torch.load(model_path, map_location=lambda storage, loc: storage)
        load_model_weight(model, ckpt, logger)
        self.model = model.to(device).eval()
        self.pipeline = Pipeline(cfg.data.val.pipeline, cfg.data.val.keep_ratio)
        
    def inference(self, img):
        img_info = {"id": 0, "file_name": os.path.basename(img) if isinstance(img, str) else None}
        img = cv2.imread(img) if isinstance(img, str) else img
        img_info.update({"height": img.shape[0], "width": img.shape[1]})
        meta = dict(img_info=img_info, raw_img=img, img=img)
        meta = self.pipeline(None, meta, self.cfg.data.val.input_size)
        meta["img"] = torch.from_numpy(meta["img"].transpose(2, 0, 1)).to(self.device)
        meta = naive_collate([meta])
        meta["img"] = stack_batch_img(meta["img"], divisible=32)
        with torch.no_grad():
            results = self.model.inference(meta)
        return meta, results
    
    def visualize(self, dets, meta, class_names):
        return self.model.head.show_result(meta["raw_img"][0], dets, class_names, score_thres=self.score_thres, show=False)

def get_image_list(path):
    return [os.path.join(maindir, filename) for maindir, _, filelist in os.walk(path) for filename in filelist if os.path.splitext(filename)[1] in image_ext] if os.path.isdir(path) else [path]

def display_crops_with_ocr(cropped_objects):
    """
    Display cropped objects with OCR text
    """
    n_crops = len(cropped_objects)
    if n_crops == 0:
        print("No objects detected!")
        return
    
    # Determine grid layout
    cols = min(3, n_crops)  # Maximum 3 columns
    rows = (n_crops + cols - 1) // cols  # Calculate needed rows
    
    # Create figure with subplots
    fig = plt.figure(figsize=(cols * 5, rows * 6))
    
    # Display each cropped object with OCR text
    for i, item in enumerate(cropped_objects):
        if len(item) == 5:  # With OCR
            crop, class_name, score, bbox, ocr_text = item
        else:  # Without OCR
            crop, class_name, score, bbox = item
            ocr_text = "OCR not performed"
        
        ax = fig.add_subplot(rows, cols, i + 1)
        ax.imshow(cv2.cvtColor(crop, cv2.COLOR_BGR2RGB))
        ax.axis('off')
        
        x0, y0, x1, y1 = bbox
        width = x1 - x0
        height = y1 - y0
        
        # Format title and OCR text
        title = f"{class_name}: {score:.2f}\nSize: {width}x{height}px"
        
        # Add title to the plot
        ax.set_title(title)
        
        # Add OCR text below the image
        if ocr_text and ocr_text != "No text found in the image." and "Error" not in ocr_text:
            # Limit OCR text to prevent overcrowding
            if len(ocr_text) > 200:
                ocr_text = ocr_text[:197] + "..."
            text_size = min(10, max(6, 10 - len(ocr_text) // 100))  # Adjust text size based on content length
            ax.text(0.5, -0.05, f"OCR: {ocr_text}", 
                    wrap=True, 
                    horizontalalignment='center', 
                    verticalalignment='top',
                    transform=ax.transAxes,
                    fontsize=text_size,
                    bbox=dict(facecolor='white', alpha=0.7))
    
    plt.tight_layout()
    plt.subplots_adjust(hspace=0.4)  # Adjust spacing between plots
    plt.show()

# Example usage
config_path = 'config/nanodet-plus-m_416-yolo.yml'
model_path = 'workspace/nanodet-plus-m_416/model_best/nanodet_model_best.pth'
image_path = 'Test/PXL_20250310_062246240.jpg'

# Initialize Azure client
try:
    azure_client = initialize_azure_client()
    print("Azure OCR client initialized successfully.")
except Exception as e:
    print(f"Warning: Azure OCR client initialization failed: {str(e)}")
    print("Continuing without OCR capability.")
    azure_client = None

# Run inference with cropping and OCR
result_images, all_cropped_objects = run_inference_for_image(
    config_path, model_path, image_path, 
    azure_client=azure_client,
    save_result=True, save_dir='Results', 
    score_thres=CONFIDENCE_THRESHOLD
)

# Display crops with OCR results
if all_cropped_objects:
    # Get the first image's crops
    image_name, cropped_objects = all_cropped_objects[0]
    
    # Display the cropped objects with OCR text
    print(f"Displaying {len(cropped_objects)} cropped objects from image: {image_name}")
    display_crops_with_ocr(cropped_objects)
    
    # Print information about detected objects
    print(f"Detected {len(cropped_objects)} objects:")
    for i, item in enumerate(cropped_objects):
        if len(item) == 5:  # With OCR
            _, class_name, score, bbox, ocr_text = item
            x0, y0, x1, y1 = bbox
            width = x1 - x0
            height = y1 - y0
            print(f"  {i+1}. Class: {class_name}, Score: {score:.2f}, Size: {width}x{height}px")
            print(f"     OCR text: {ocr_text}")
        else:  # Without OCR
            _, class_name, score, bbox = item
            x0, y0, x1, y1 = bbox
            width = x1 - x0
            height = y1 - y0
            print(f"  {i+1}. Class: {class_name}, Score: {score:.2f}, Size: {width}x{height}px")
else:
    print("No result images available.")
```

```python
import os
import time
import cv2
import torch
import numpy as np
import matplotlib.pyplot as plt
from nanodet.data.batch_process import stack_batch_img
from nanodet.data.collate import naive_collate
from nanodet.data.transform import Pipeline
from nanodet.model.arch import build_model
from nanodet.util import Logger, cfg, load_config, load_model_weight
from nanodet.util.path import mkdir
from azure.cognitiveservices.vision.computervision import ComputerVisionClient
from azure.cognitiveservices.vision.computervision.models import OperationStatusCodes
from msrest.authentication import CognitiveServicesCredentials
from io import BytesIO
import re

image_ext = [".jpg", ".jpeg", ".webp", ".bmp", ".png"]
# Confidence Threshold for Detection
CONFIDENCE_THRESHOLD = 0.50

# Initialize Azure OCR Client
def initialize_azure_client():
    """Initialize and return Azure Computer Vision client"""
    api_key = os.getenv("AZURE_API_KEY")
    endpoint = os.getenv("AZURE_ENDPOINT")
    if not api_key or not endpoint:
        raise ValueError("Azure API Key or Endpoint is missing! Set environment variables properly.")
    return ComputerVisionClient(endpoint, CognitiveServicesCredentials(api_key))

def extract_license_plate_text(ocr_text):
    """
    Extract only the license plate number from OCR text,
    ignoring country codes typically on the left side.
    
    Args:
        ocr_text: Raw OCR text from Azure
        
    Returns:
        Cleaned license plate number text
    """
    if not ocr_text or "Error" in ocr_text or "No text" in ocr_text:
        return ocr_text
    
    # Split by newlines to handle multi-line OCR results
    lines = ocr_text.strip().split('\n')
    
    # Process each line
    cleaned_lines = []
    for line in lines:
        # Pattern to match license plate format (adjust as needed for your region)
        # This pattern looks for alphanumeric sequences with common license plate formats
        # It will ignore short country codes (1-3 characters) at the beginning
        plate_pattern = r'([A-Z]{2}\s*[0-9]{1,2}\s*[A-Z]{1,2}\s*[0-9]{4})'
        
        match = re.search(plate_pattern, line)
        if match:
            # Extract the main portion of the license plate number
            plate_number = match.group(1)
            cleaned_lines.append(plate_number.strip())
    
    # Return the joined result or original if nothing matched
    if cleaned_lines:
        return ' '.join(cleaned_lines)
    return "No valid license plate format detected"

def extract_text_from_image(client, image):
    """
    Extracts text from an image using Azure OCR and filters for license plate text.
    
    Args:
        client: Azure Computer Vision client
        image: Image as numpy array (BGR format)
    
    Returns:
        Extracted license plate text as string
    """
    try:
        # Convert numpy array to bytes
        is_success, buffer = cv2.imencode(".jpg", image)
        if not is_success:
            return "Error: Failed to encode image."
        
        image_bytes = buffer.tobytes()
        
        # Convert bytes to BytesIO object (file-like object)
        image_stream = BytesIO(image_bytes)
        
        # Send image to Azure OCR
        response = client.read_in_stream(image_stream, raw=True)
        
        # Check if response is None
        if response is None:
            return "Error: No response received from Azure OCR API."
        
        # Extract operation ID correctly
        operation_location = response.headers.get("Operation-Location")
        if not operation_location:
            return "Error: Operation-Location header missing in response."
        
        operation_id = operation_location.split("/")[-1]
        
        # Wait for OCR to complete (Max 30s timeout)
        max_wait_time = 30  # Max wait time in seconds
        start_time = time.time()
        while True:
            result = client.get_read_result(operation_id)
            if result.status in [OperationStatusCodes.succeeded, OperationStatusCodes.failed]:
                break
            if time.time() - start_time > max_wait_time:
                return "Error: OCR processing timed out."
            time.sleep(2)  # Delay to avoid excessive API calls
        
        # Extract and return text
        extracted_text = []
        if result.status == OperationStatusCodes.succeeded:
            for page in result.analyze_result.read_results:
                for line in page.lines:
                    extracted_text.append(line.text)
        
        raw_text = "\n".join(extracted_text) if extracted_text else "No text found in the image."
        
        # Filter for license plate text if this is a license plate
        if raw_text and "No text" not in raw_text and "Error" not in raw_text:
            return extract_license_plate_text(raw_text)
        return raw_text
    
    except Exception as e:
        return f"Error: {str(e)}"

def crop_detected_objects(image, dets, class_names, score_thres=0.5):
    """
    Crop and return detected objects from the image
    Args:
        image: original image
        dets: detection results
        class_names: list of class names
        score_thres: score threshold for filtering detections
    Returns:
        List of tuples (cropped_image, class_name, score, bbox)
    """
    cropped_objects = []
    
    for label in range(len(class_names)):
        for bbox in dets[label]:
            score = bbox[-1]
            if score > score_thres:
                x0, y0, x1, y1 = [int(i) for i in bbox[:4]]
                # Ensure box is within image boundaries
                x0 = max(0, x0)
                y0 = max(0, y0)
                x1 = min(image.shape[1], x1)
                y1 = min(image.shape[0], y1)
                
                # Crop the object
                cropped = image[y0:y1, x0:x1]
                
                if cropped.size > 0:  # Check if crop is valid
                    cropped_objects.append((
                        cropped, 
                        class_names[label], 
                        score, 
                        (x0, y0, x1, y1)
                    ))
    
    return cropped_objects

def run_inference_for_image(config_path, model_path, image_path, azure_client=None, save_result=False, save_dir='./demo_results', score_thres=0.5):
    load_config(cfg, config_path)
    logger = Logger(local_rank=0, use_tensorboard=False)
    predictor = Predictor(cfg, model_path, logger, device="cuda:0", score_thres=score_thres)
    
    image_names = get_image_list(image_path)
    image_names.sort()
    if save_result:
        save_folder = os.path.join(save_dir, time.strftime("%Y_%m_%d_%H_%M_%S", time.localtime()))
        mkdir(local_rank=0, path=save_folder)
        
        # Create subfolder for crops
        crop_folder = os.path.join(save_folder, "crops")
        mkdir(local_rank=0, path=crop_folder)
    
    result_images = []
    all_cropped_objects = []
    
    for image_name in image_names:
        meta, res = predictor.inference(image_name)
        raw_img = meta["raw_img"][0]
        result_image = predictor.visualize(res[0], meta, cfg.class_names)
        
        # Crop detected objects
        cropped_objects = crop_detected_objects(raw_img, res[0], cfg.class_names, score_thres)
        
        # Process OCR if Azure client is provided
        if azure_client:
            cropped_objects_with_ocr = []
            for crop, class_name, score, bbox in cropped_objects:
                ocr_text = extract_text_from_image(azure_client, crop)
                cropped_objects_with_ocr.append((
                    crop, 
                    class_name, 
                    score, 
                    bbox, 
                    ocr_text
                ))
            all_cropped_objects.append((os.path.basename(image_name), cropped_objects_with_ocr))
        else:
            all_cropped_objects.append((os.path.basename(image_name), cropped_objects))
        
        if save_result:
            # Save the image with bounding boxes
            save_file_name = os.path.join(save_folder, os.path.basename(image_name))
            cv2.imwrite(save_file_name, result_image)
            
            # Save cropped objects
            for i, item in enumerate(all_cropped_objects[-1][1]):
                if azure_client:
                    crop, class_name, score, _, ocr_text = item
                    # Save OCR text to file
                    text_filename = f"{os.path.splitext(os.path.basename(image_name))[0]}_{class_name}_{i}_{score:.2f}.txt"
                    text_path = os.path.join(crop_folder, text_filename)
                    with open(text_path, 'w') as f:
                        f.write(ocr_text)
                else:
                    crop, class_name, score, _ = item
                
                crop_filename = f"{os.path.splitext(os.path.basename(image_name))[0]}_{class_name}_{i}_{score:.2f}.jpg"
                crop_path = os.path.join(crop_folder, crop_filename)
                cv2.imwrite(crop_path, crop)
        
        result_images.append(result_image)
    
    return result_images, all_cropped_objects

class Predictor:
    def __init__(self, cfg, model_path, logger, device="cuda:0", score_thres=0.5):
        self.cfg = cfg
        self.device = device
        self.score_thres = score_thres
        model = build_model(cfg.model)
        ckpt = torch.load(model_path, map_location=lambda storage, loc: storage)
        load_model_weight(model, ckpt, logger)
        self.model = model.to(device).eval()
        self.pipeline = Pipeline(cfg.data.val.pipeline, cfg.data.val.keep_ratio)
        
    def inference(self, img):
        img_info = {"id": 0, "file_name": os.path.basename(img) if isinstance(img, str) else None}
        img = cv2.imread(img) if isinstance(img, str) else img
        img_info.update({"height": img.shape[0], "width": img.shape[1]})
        meta = dict(img_info=img_info, raw_img=img, img=img)
        meta = self.pipeline(None, meta, self.cfg.data.val.input_size)
        meta["img"] = torch.from_numpy(meta["img"].transpose(2, 0, 1)).to(self.device)
        meta = naive_collate([meta])
        meta["img"] = stack_batch_img(meta["img"], divisible=32)
        with torch.no_grad():
            results = self.model.inference(meta)
        return meta, results
    
    def visualize(self, dets, meta, class_names):
        return self.model.head.show_result(meta["raw_img"][0], dets, class_names, score_thres=self.score_thres, show=False)

def get_image_list(path):
    return [os.path.join(maindir, filename) for maindir, _, filelist in os.walk(path) for filename in filelist if os.path.splitext(filename)[1] in image_ext] if os.path.isdir(path) else [path]

def display_crops_with_ocr(cropped_objects):
    """
    Display cropped objects with OCR text
    """
    n_crops = len(cropped_objects)
    if n_crops == 0:
        print("No objects detected!")
        return
    
    # Determine grid layout
    cols = min(3, n_crops)  # Maximum 3 columns
    rows = (n_crops + cols - 1) // cols  # Calculate needed rows
    
    # Create figure with subplots
    fig = plt.figure(figsize=(cols * 5, rows * 6))
    
    # Display each cropped object with OCR text
    for i, item in enumerate(cropped_objects):
        if len(item) == 5:  # With OCR
            crop, class_name, score, bbox, ocr_text = item
        else:  # Without OCR
            crop, class_name, score, bbox = item
            ocr_text = "OCR not performed"
        
        ax = fig.add_subplot(rows, cols, i + 1)
        ax.imshow(cv2.cvtColor(crop, cv2.COLOR_BGR2RGB))
        ax.axis('off')
        
        x0, y0, x1, y1 = bbox
        width = x1 - x0
        height = y1 - y0
        
        # Format title and OCR text
        title = f"{class_name}: {score:.2f}\nSize: {width}x{height}px"
        
        # Add title to the plot
        ax.set_title(title)
        
        # Add OCR text below the image
        if ocr_text and ocr_text != "No text found in the image." and "Error" not in ocr_text:
            # Limit OCR text to prevent overcrowding
            if len(ocr_text) > 200:
                ocr_text = ocr_text[:197] + "..."
            text_size = min(10, max(6, 10 - len(ocr_text) // 100))  # Adjust text size based on content length
            ax.text(0.5, -0.05, f"OCR: {ocr_text}", 
                    wrap=True, 
                    horizontalalignment='center', 
                    verticalalignment='top',
                    transform=ax.transAxes,
                    fontsize=text_size,
                    bbox=dict(facecolor='white', alpha=0.7))
    
    plt.tight_layout()
    plt.subplots_adjust(hspace=0.4)  # Adjust spacing between plots
    plt.show()

# Example usage
config_path = 'config/nanodet-plus-m_416-yolo.yml'
model_path = 'workspace/nanodet-plus-m_416/model_best/nanodet_model_best.pth'
image_path = 'Test/PXL_20250310_062238801.jpg'

# Initialize Azure client
try:
    azure_client = initialize_azure_client()
    print("Azure OCR client initialized successfully.")
except Exception as e:
    print(f"Warning: Azure OCR client initialization failed: {str(e)}")
    print("Continuing without OCR capability.")
    azure_client = None

# Run inference with cropping and OCR
result_images, all_cropped_objects = run_inference_for_image(
    config_path, model_path, image_path, 
    azure_client=azure_client,
    save_result=True, save_dir='Results', 
    score_thres=CONFIDENCE_THRESHOLD
)

# Display crops with OCR results
if all_cropped_objects:
    # Get the first image's crops
    image_name, cropped_objects = all_cropped_objects[0]
    
    # Display the cropped objects with OCR text
    print(f"Displaying {len(cropped_objects)} cropped objects from image: {image_name}")
    display_crops_with_ocr(cropped_objects)
    
    # Print information about detected objects
    print(f"Detected {len(cropped_objects)} objects:")
    for i, item in enumerate(cropped_objects):
        if len(item) == 5:  # With OCR
            _, class_name, score, bbox, ocr_text = item
            x0, y0, x1, y1 = bbox
            width = x1 - x0
            height = y1 - y0
            print(f"  {i+1}. Class: {class_name}, Score: {score:.2f}, Size: {width}x{height}px")
            print(f"     OCR text: {ocr_text}")
        else:  # Without OCR
            _, class_name, score, bbox = item
            x0, y0, x1, y1 = bbox
            width = x1 - x0
            height = y1 - y0
            print(f"  {i+1}. Class: {class_name}, Score: {score:.2f}, Size: {width}x{height}px")
else:
    print("No result images available.")
```

```python
import os
import time
import cv2
import torch
import numpy as np
import matplotlib.pyplot as plt
from nanodet.data.batch_process import stack_batch_img
from nanodet.data.collate import naive_collate
from nanodet.data.transform import Pipeline
from nanodet.model.arch import build_model
from nanodet.util import Logger, cfg, load_config, load_model_weight
from nanodet.util.path import mkdir
from azure.cognitiveservices.vision.computervision import ComputerVisionClient
from azure.cognitiveservices.vision.computervision.models import OperationStatusCodes
from msrest.authentication import CognitiveServicesCredentials
from io import BytesIO
import re

image_ext = [".jpg", ".jpeg", ".webp", ".bmp", ".png"]
CONFIDENCE_THRESHOLD = 0.50

def initialize_azure_client():
    api_key = os.getenv("AZURE_API_KEY")
    endpoint = os.getenv("AZURE_ENDPOINT")
    if not api_key or not endpoint:
        raise ValueError("Azure API Key or Endpoint is missing!")
    return ComputerVisionClient(endpoint, CognitiveServicesCredentials(api_key))

def extract_license_plate_text(ocr_text):
    if not ocr_text or "Error" in ocr_text or "No text" in ocr_text:
        return ocr_text
    lines = ocr_text.strip().split('\n')
    cleaned_lines = []
    plate_pattern = r'([A-Z]{2}\s*[0-9]{1,2}\s*[A-Z]{1,2}\s*[0-9]{4})'
    for line in lines:
        match = re.search(plate_pattern, line)
        if match:
            cleaned_lines.append(match.group(1).strip())
    return ' '.join(cleaned_lines) if cleaned_lines else "No valid license plate format detected"

def extract_text_from_image(client, image):
    try:
        _, buffer = cv2.imencode(".jpg", image)
        image_stream = BytesIO(buffer.tobytes())
        response = client.read_in_stream(image_stream, raw=True)
        operation_id = response.headers.get("Operation-Location").split("/")[-1]
        max_wait_time = 30
        start_time = time.time()
        while True:
            result = client.get_read_result(operation_id)
            if result.status in [OperationStatusCodes.succeeded, OperationStatusCodes.failed]:
                break
            if time.time() - start_time > max_wait_time:
                return "Error: OCR processing timed out."
            time.sleep(2)
        extracted_text = "\n".join(line.text for page in result.analyze_result.read_results for line in page.lines) if result.status == OperationStatusCodes.succeeded else "No text found."
        return extract_license_plate_text(extracted_text) if extracted_text else extracted_text
    except Exception as e:
        return f"Error: {str(e)}"

def crop_detected_objects(image, dets, class_names, score_thres=0.5):
    return [(image[int(y0):int(y1), int(x0):int(x1)], class_names[label], score, (x0, y0, x1, y1))
            for label in range(len(class_names))
            for bbox in dets[label] if (score := bbox[-1]) > score_thres
            for x0, y0, x1, y1 in [bbox[:4]]]

def run_inference_for_image(config_path, model_path, image_path, azure_client=None, save_result=False, save_dir='./demo_results', score_thres=0.5):
    load_config(cfg, config_path)
    logger = Logger(local_rank=0, use_tensorboard=False)
    predictor = Predictor(cfg, model_path, logger, device="cuda:0", score_thres=score_thres)
    image_names = sorted(get_image_list(image_path))
    
    save_folder = os.path.join(save_dir, time.strftime("%Y_%m_%d_%H_%M_%S", time.localtime())) if save_result else None
    mkdir(local_rank=0, path=save_folder) if save_result else None
    crop_folder = os.path.join(save_folder, "crops") if save_result else None
    mkdir(local_rank=0, path=crop_folder) if save_result else None

    all_cropped_objects = []
    result_images = []  # Initialize result_images here

    for image_name in image_names:
        meta, res = predictor.inference(image_name)
        raw_img = meta["raw_img"][0]
        result_image = predictor.visualize(res[0], meta, cfg.class_names)
        cropped_objects = crop_detected_objects(raw_img, res[0], cfg.class_names, score_thres)

        if azure_client:
            cropped_objects_with_ocr = [(crop, class_name, score, bbox, extract_text_from_image(azure_client, crop)) for crop, class_name, score, bbox in cropped_objects]
            all_cropped_objects.append((os.path.basename(image_name), cropped_objects_with_ocr))
        else:
            all_cropped_objects.append((os.path.basename(image_name), cropped_objects))

        if save_result:
            cv2.imwrite(os.path.join(save_folder, os.path.basename(image_name)), result_image)
            for i, item in enumerate(all_cropped_objects[-1][1]):
                crop, class_name, score, _, ocr_text = item if azure_client else item + ("OCR not performed",)
                with open(os.path.join(crop_folder, f"{os.path.splitext(os.path.basename(image_name))[0]}_{class_name}_{i}_{score:.2f}.txt"), 'w') as f:
                    f.write(ocr_text)
                cv2.imwrite(os.path.join(crop_folder, f"{os.path.splitext(os.path.basename(image_name))[0]}_{class_name}_{i}_{score:.2f}.jpg"), crop)

        result_images.append(result_image)  # Append the result image to the list

    return result_images, all_cropped_objects

class Predictor:
    def __init__(self, cfg, model_path, logger, device="cuda:0", score_thres=0.5):
        self.cfg = cfg
        self.device = device
        self.score_thres = score_thres
        model = build_model(cfg.model)
        load_model_weight(model, torch.load(model_path, map_location=lambda storage, loc: storage), logger)
        self.model = model.to(device).eval()
        self.pipeline = Pipeline(cfg.data.val.pipeline, cfg.data.val.keep_ratio)

    def inference(self, img):
        img_info = {"id": 0, "file_name": os.path.basename(img) if isinstance(img, str) else None}
        img = cv2.imread(img) if isinstance(img, str) else img
        img_info.update({"height": img.shape[0], "width": img.shape[1]})
        meta = dict(img_info=img_info, raw_img=img, img=img)
        meta = self.pipeline(None, meta, self.cfg.data.val.input_size)
        meta["img"] = torch.from_numpy(meta["img"].transpose(2, 0, 1)).to(self.device)
        meta = naive_collate([meta])
        meta["img"] = stack_batch_img(meta["img"], divisible=32)
        with torch.no_grad():
            return meta, self.model.inference(meta)

    def visualize(self, dets, meta, class_names):
        return self.model.head.show_result(meta["raw_img"][0], dets, class_names, score_thres=self.score_thres, show=False)

def get_image_list(path):
    return [os.path.join(maindir, filename) for maindir, _, filelist in os.walk(path) for filename in filelist if os.path.splitext(filename)[1] in image_ext] if os.path.isdir(path) else [path]

def display_crops_with_ocr(cropped_objects):
    if not cropped_objects:
        print("No objects detected!")
        return
    cols, rows = min(3, len(cropped_objects)), (len(cropped_objects) + 2) // 3
    fig = plt.figure(figsize=(cols * 5, rows * 6))
    for i, item in enumerate(cropped_objects):
        crop, class_name, score, bbox, ocr_text = item if len(item) == 5 else (*item, "OCR not performed")
        ax = fig.add_subplot(rows, cols, i + 1)
        ax.imshow(cv2.cvtColor(crop, cv2.COLOR_BGR2RGB))
        ax.axis('off')
        ax.set_title(f"{class_name}: {score:.2f}\nSize: {bbox[2]-bbox[0]}x{bbox[3]-bbox[1]}px")
        ax.text(0.5, -0.05, f"OCR: {ocr_text[:197]}...", wrap=True, horizontalalignment='center', verticalalignment='top', transform=ax.transAxes, fontsize=min(10, max(6, 10 - len(ocr_text) // 100)), bbox=dict(facecolor='white', alpha=0.7))
    plt.tight_layout()
    plt.subplots_adjust(hspace=0.4)
    plt.show()

# Example usage
config_path = 'config/nanodet-plus-m_416-yolo.yml'
model_path = 'workspace/nanodet-plus-m_416/model_best/nanodet_model_best.pth'
image_path = 'Test/PXL_20250310_062246240.jpg'

try:
    azure_client = initialize_azure_client()
    print("Azure OCR client initialized successfully.")
except Exception as e:
    print(f"Warning: Azure OCR client initialization failed: {str(e)}")
    azure_client = None

result_images, all_cropped_objects = run_inference_for_image(
    config_path, model_path, image_path, 
    azure_client=azure_client,
    save_result=True, save_dir='Results', 
    score_thres=CONFIDENCE_THRESHOLD
)

if all_cropped_objects:
    image_name, cropped_objects = all_cropped_objects[0]
    print(f"Displaying {len(cropped_objects)} cropped objects from image: {image_name}")
    display_crops_with_ocr(cropped_objects)
    for i, item in enumerate(cropped_objects):
        if len(item) == 5:
            _, class_name, score, bbox, ocr_text = item
        else:
            _, class_name, score, bbox = item
        print(f"  {i+1}. Class: {class_name}, Score: {score:.2f}, Size: {bbox[2]-bbox[0]}x{bbox[3]-bbox[1]}px")
        print(f"     OCR text: {ocr_text if len(item) == 5 else 'OCR not performed'}")
else:
    print("No result images available.")
```
